```python
NUM_CARS = 5
print(f"--- Starting Car Entry Process for {NUM_CARS} Cars ---")

# List to store all car data
all_car_summaries = []

# Totals
total_msrp = 0
total_cost = 0
total_gross_profit = 0
total_adjusted_profit = 0

for i in range(NUM_CARS):
    print(f"\n--- Entering Data for Car #{i + 1} ---")

    # Inputs
    Car_Name = input("Car name: ")
    MSRP = float(input("MSRP: "))
    Cost = float(input("Cost: "))
    Car_Type_Input = input("Type ('domestic' or 'import'): ").lower()

    # Calculations
    Potential_Gross_Profit = MSRP - Cost
    Adjusted_Potential_Gross_Profit = Potential_Gross_Profit

    if Car_Type_Input == "import":
        reduction = 0.0175
        Adjusted_Potential_Gross_Profit -= Potential_Gross_Profit * reduction

    Car_Type = Car_Type_Input if Car_Type_Input in ["domestic", "import"] else "Unknown"

    # Store car summary
    car_summary = {
        "Name": Car_Name,
        "Type": Car_Type,
        "MSRP": MSRP,
        "Cost": Cost,
        "Gross_Profit": Potential_Gross_Profit,
        "Adjusted_Profit": Adjusted_Potential_Gross_Profit
    }
    all_car_summaries.append(car_summary)

    # Update totals
    total_msrp += MSRP
    total_cost += Cost
    total_gross_profit += Potential_Gross_Profit
    total_adjusted_profit += Adjusted_Potential_Gross_Profit

# Final Summary
print("\n--- All Cars Processed ---")
print("\n=== Summary of All Cars ===")
for idx, car in enumerate(all_car_summaries, start=1):
    print(f"\nCar #{idx}: {car['Name']}")
    print(f"Type: {car['Type']}")
    print(f"MSRP: ${car['MSRP']}")
    print(f"Cost: ${car['Cost']}")
    print(f"Gross Profit: ${car['Gross_Profit']}")
    print(f"Adjusted Profit: ${car['Adjusted_Profit']}")

print("\n=== Totals ===")
print(f"Total MSRP: ${total_msrp}")
print(f"Total Cost: ${total_cost}")
print(f"Total Gross Profit: ${total_gross_profit}")
print(f"Total Adjusted Gross Profit: ${total_adjusted_profit}")
```
